```python
# ==========================================================
# Experiment 1: Single Layer Perceptron
# Objective:
# Implement AND, OR, and NOT logic gates using
# Single Layer Perceptron from scratch.
#
# Tools:
# Python, NumPy, Matplotlib
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt

# ----------------------------------------------------------
# Step Activation Function
# ----------------------------------------------------------
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0


# ----------------------------------------------------------
# Perceptron Class
# ----------------------------------------------------------
class Perceptron:

    def __init__(self, learning_rate=0.1, epochs=20):
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X, y):

        # Initialize weights and bias with zeros
        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        print("="*50)
        print("Training Started")
        print("="*50)

        # Training Loop
        for epoch in range(self.epochs):

            print(f"\nEpoch {epoch+1}")

            for i in range(len(X)):

                # Linear Combination
                linear_output = np.dot(X[i], self.weights) + self.bias

                # Prediction
                prediction = step_function(linear_output)

                # Error
                error = y[i] - prediction

                # Weight Update
                self.weights = self.weights + self.lr * error * X[i]

                # Bias Update
                self.bias = self.bias + self.lr * error

                print("Input:", X[i],
                      " Target:", y[i],
                      " Prediction:", prediction,
                      " Error:", error)

            print("Weights:", self.weights)
            print("Bias:", self.bias)

        print("\nTraining Completed")
        print("="*50)

    def predict(self, X):

        predictions = []

        for sample in X:

            linear_output = np.dot(sample, self.weights) + self.bias
            predictions.append(step_function(linear_output))

        return np.array(predictions)


# ----------------------------------------------------------
# Plot Decision Boundary
# ----------------------------------------------------------
def plot_decision_boundary(model, X, y, title):

    plt.figure(figsize=(6,5))

    for i in range(len(X)):
        if y[i] == 0:
            plt.scatter(X[i][0], X[i][1],
                        color='red',
                        marker='o',
                        s=100)

        else:
            plt.scatter(X[i][0], X[i][1],
                        color='blue',
                        marker='^',
                        s=100)

    x_values = np.linspace(-0.5, 1.5, 100)

    if model.weights[1] != 0:
        y_values = -(model.weights[0] * x_values + model.bias) / model.weights[1]
        plt.plot(x_values, y_values, color='green', linewidth=2)

    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)
    plt.grid(True)

    plt.title(title)
    plt.xlabel("Input X1")
    plt.ylabel("Input X2")

    plt.show()


# ==========================================================
# AND Gate
# ==========================================================

print("\n\n==============================")
print("AND GATE")
print("==============================")

X_and = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y_and = np.array([0,0,0,1])

and_model = Perceptron(learning_rate=0.1, epochs=10)
and_model.fit(X_and, y_and)

prediction = and_model.predict(X_and)

print("\nFinal Predictions")
print(prediction)

plot_decision_boundary(and_model,
                       X_and,
                       y_and,
                       "AND Gate Decision Boundary")


# ==========================================================
# OR Gate
# ==========================================================

print("\n\n==============================")
print("OR GATE")
print("==============================")

X_or = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y_or = np.array([0,1,1,1])

or_model = Perceptron(learning_rate=0.1, epochs=10)
or_model.fit(X_or, y_or)

prediction = or_model.predict(X_or)

print("\nFinal Predictions")
print(prediction)

plot_decision_boundary(or_model,
                       X_or,
                       y_or,
                       "OR Gate Decision Boundary")


# ==========================================================
# NOT Gate
# ==========================================================

print("\n\n==============================")
print("NOT GATE")
print("==============================")

X_not = np.array([
    [0],
    [1]
])

y_not = np.array([1,0])

not_model = Perceptron(learning_rate=0.1, epochs=10)
not_model.fit(X_not, y_not)

prediction = not_model.predict(X_not)

print("\nFinal Predictions")
print(prediction)


# ==========================================================
# Testing User Inputs
# ==========================================================

print("\nTesting AND Gate")

test = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

print(and_model.predict(test))

print("\nTesting OR Gate")

print(or_model.predict(test))

print("\nTesting NOT Gate")

print(not_model.predict(np.array([[0],[1]])))
```
